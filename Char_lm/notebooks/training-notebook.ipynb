```python
import shutil
import os
import sys

# --- ⚠️ Set the Source Path ⚠️ ---
# The source folder you want to copy
SOURCE_FOLDER_PATH = "/kaggle/input/transformer43/mini-transformer-from-scratch"

# The base destination path (always /kaggle/working/ on Kaggle)
KAGGLE_WORKING_PATH = "/kaggle/working"
# ------------------------------------

def robust_kaggle_folder_copy(source_path, destination_base_path):
    """
    Copies a folder by creating a new subfolder in the destination, 
    making it safe for use with shutil.copytree() in Kaggle.
    """
    
    # 1. Extract the name of the folder being copied (e.g., 'mini-transformer-from-scratch')
    folder_name = os.path.basename(source_path)
    
    # 2. Construct the full new destination path
    # This ensures copytree creates a new directory inside /kaggle/working/
    full_destination_path = os.path.join(destination_base_path, folder_name)
    
    print(f"Starting copy:\n  FROM: {source_path}\n  TO:   {full_destination_path}")

    # Check if the source exists
    if not os.path.isdir(source_path):
        print(f"❌ Error: Source folder not found at '{source_path}'")
        sys.exit(1)

    try:
        # Check if the intended destination subfolder already exists
        if os.path.exists(full_destination_path):
            print(f"⚠️ Warning: Destination subfolder '{full_destination_path}' already exists. Deleting it...")
            # Delete the existing folder before copying
            shutil.rmtree(full_destination_path)
            print("   Existing folder deleted.")

        # 3. Perform the copy operation
        # shutil.copytree will now safely create the new folder 'mini-transformer-from-scratch'
        shutil.copytree(source_path, full_destination_path)
        
        print("\n✅ Success! Folder copied completely.")
        
    except Exception as e:
        print(f"❌ An unexpected error occurred: {e}")
        sys.exit(1)

if __name__ == "__main__":
    # If you want the script to be idempotent (safe to run multiple times), 
    # use the robust version that deletes the existing target.
    robust_kaggle_folder_copy(SOURCE_FOLDER_PATH, KAGGLE_WORKING_PATH)
```
